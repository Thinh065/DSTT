1. Dẫn nhập – Một số hàm về xử lý vector với Python 

```python
def scale(a, v):
 return [a*vi for vi in v]
v = [3,5,7]
scale(10, v) 
```

```python
def sumvector(v, w):
 return [vi+wi for (vi, wi) in zip(v, w)]
v = [3,5,7]
w = [2,4,6]
sumvector(v, w)
```

```python
def dotvector(v, w): 
    return sum([vi*wi for (vi, wi) in zip(v, w)])
dotvector(v, w) 
print(dotvector(v, w))
```

```python
def lenvector(v):
 return dotvector(v,v)
lenvector(w) 
```

4. Bài toán ứng dụng 2 – Tính toán dãy Fibonacci: Con đường tìm đến tỉ số vàng! 

```python
import numpy as np

A = np.array([[1, 1],
              [1, 0]])

b = np.array([1, 0])

n = 10
for i in range(n):
    b = A.dot(b)
    print(b)
```

BÀI TẬP CHƯƠNG 2 


```python
# B1
import sympy as sp

x = sp.symbols('x')
a, b, c = sp.symbols('a b c')

# Biểu thức
R = a/(x-1) + b/(x-1)**2 + c/(x+2)

# Quy đồng
R_simplified = sp.simplify(R)
print(R_simplified)

# So sánh với dạng chuẩn (tử số)
target = ((a+c)*x**2 + (a + b - 2*c)*x + (-2*a + 2*b + c)) / ((x-1)**2 * (x+2))

print(sp.simplify(R_simplified - target))  # = 0 nếu đúng
```

```python
# Bài 2: Giải hệ phương trình tìm a, b, c
solutions = sp.solve([
    a + c - 1,
    a + b - 2*c + 3
], (a, b, c))

print(solutions)
```

```python
# Bài 3: Kiểm tra lại kết quả
solutions = sp.solve([
    a + c - 1,
    a + b - 2*c + 3
], (a, b, c), dict=True)

sol = solutions[0]   # lúc này OK
print(sol)
```
