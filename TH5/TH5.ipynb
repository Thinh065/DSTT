2.2. Liên phân số biểu diễn số pi

```python
import numpy as np

c = [3, 7, 15, 1, 292, 1, 1, 1, 2, 1, 3, 1, 14, 2]

M = np.array([[1, 0],
              [0, 1]])  # ma trận đơn vị 2x2

for i in range(len(c)):
    ci = np.array([[c[i], 1],
                   [1, 0]])

    M = M @ ci
    print(M)

print(M[0,0] / M[1,0])
```

3. Về phân tích hồi quy tuyến tính

3.1. Dẫn nhập: khái niệm về sai số

3.2. Phương pháp bình phương cực tiểu (mô hình tuyến tính)

```python
X = np.array([[1,1],
              [1,2],
              [1,3],
              [1,4],
              [1,5]])

Y = np.array([[1],
              [2],
              [4],
              [4],
              [6]])

XT = X.T

A1 = np.linalg.inv(XT @ X)
A2 = XT @ Y

A = A1 @ A2
print(A)
```

```python
import numpy as np

def bpcuctieu(a):
    b = []
    o1 = []
    for i in range(len(a)):
        o1.append(1.0)
        # Theo code của bạn: i+1 làm x chạy từ 1 đến 5
        b.append((i+1)*1.0) 
    
    # Xây dựng ma trận thiết kế (Design Matrix)
    x = np.array([o1, b]) 
    
    # Tính ma trận hiệp phương sai và nghịch đảo
    u = np.dot(x, x.T) 
    u_1 = np.linalg.inv(u)
    
    # Tính toán vector hệ số A
    y = np.array(a)
    v = np.dot(u_1, x)
    A = np.dot(v, y)
    
    return A

# Chạy thử với dữ liệu trong ảnh
dulieu = [10.0, 8, 7, 5, 2]
ket_qua = bpcuctieu(dulieu)

print(f"Hệ số b (Intercept): {ket_qua[0]:.1f}")
print(f"Hệ số a (Slope): {ket_qua[1]:.1f}")
print(f"Phương trình: y = {ket_qua[0]:.1f} {ket_qua[1]:.1f}x")
```
