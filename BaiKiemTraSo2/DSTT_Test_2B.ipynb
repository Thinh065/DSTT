## **Bài 1** *(4 điểm)*

Xét các vector sau trong không gian $\mathbb{R}^3$
$$a_1 = (-1,  3, -2),\quad a_2 = (6, 1, 4),\quad a_3 = (6, 3, 1)$$
Tìm các số thực $x_1, x_2, x_3$ sao cho có thể biểu diễn vector $b = (6,  8, -1)$ dưới dạng
$$b = x_1 a_1 + x_2 a_2 + x_3 a_3$$

```python
### CODE Ở ĐÂY
import numpy as np

# Ma trận hệ số
A = np.array([
    [-1, 6, 6],
    [3, 1, 3],
    [-2, 4, 1]
])

# Vector b
b = np.array([6, 8, -1])

# Giải hệ
solution = np.linalg.solve(A, b)

x_1 = solution[0]
x_2 = solution[1]
x_3 = solution[2]

print(f'Các hệ số cần tìm: x_1 = {x_1}, x_2 = {x_2}, x_3 = {x_3}')

b_check = A @ solution

print("Vector sau khi kiểm tra lại:", b_check)
```

## **Bài 2** *(3 điểm)*

Xét các vector sau trong không gian $\mathbb{R}^4$
$$a_1 = (5, -2,  9, -3),\quad a_2 = (0, -2, -2,  1)$$
$$a_3 = (9, -1,  6, -3),\quad a_4 = (4, 2, 4, 9)$$
Tìm các số thực $x_1, x_2, x_3, x_4$ sao cho có thể biểu diễn vector $b = (4, -3, -4, 7)$ dưới dạng
$$b = x_1 a_1 + x_2 a_2 + x_3 a_3 + x_4 a_4$$

```python
import numpy as np

### CODE Ở ĐÂY

# Khai báo các vector
a1 = np.array([5, -2, 9, -3])
a2 = np.array([0, -2, -2, 1])
a3 = np.array([9, -1, 6, -3])
a4 = np.array([4, 2, 4, 9])

b = np.array([4, -3, -4, 7])

# Tạo ma trận A (các cột là a1, a2, a3, a4)
A = np.column_stack((a1, a2, a3, a4))

# Giải hệ
x = np.linalg.solve(A, b)

### Xuất kết quả
x_1 = x[0]
x_2 = x[1]
x_3 = x[2]
x_4 = x[3]

print(f'Các hệ số cần tìm: x_1 = {x_1}, x_2 = {x_2}, x_3 = {x_3}, x_4 = {x_4}')

# Kiểm tra lại
check = x_1*a1 + x_2*a2 + x_3*a3 + x_4*a4
print("Thử lại:", check)
```

## **Bài 3** *(3 điểm)*

Để tìm giá trị riêng lớn nhất và một vector riêng tương ứng của một ma trận $A \in \mathbb{R}^{n \times n}$ cho trước, ta có thể sử dụng **power method** như sau:
-  Bước 1: Chọn một vector $x_0 \in \mathbb{R}^n$ khác không, ví dụ chọn $x_0 = (1, 1, \ldots, 1)$
-  Bước 2: Lặp ($k = 1, 2, 3,\ldots$) cho đến khi hội tụ
    -  Bước 2.1: Tính tích ma trận - vector $A x$
    $$y_k = A x_k$$
    - Bước 2.2: Tính chuẩn 2 của vector $A x$
    $$\lambda_k = \|y_k\|_2$$
    - Bước 2.3: Chuẩn hóa để vector có chiều dài bằng 1
    $$ x_k = \frac{y_k}{\lambda_k}$$

Dãy $\{\lambda_k\}$ sẽ hội tụ về giá trị riêng lớn nhất của $A$, trong khi dãy $\{x_k\}$ hội tụ về vector riêng tương ứng.

Xét ma trận
$$A = \left[\begin{array}{rrrr}
-4 &  7 &  8 &  1 \\
-2 & -1 & -4 &  9 \\
-4 & -2 &  6 &  6 \\
9 &  6 &  9 &  3
\end{array}\right]$$
Sử dụng power method với $x_0 = (1, 1, 1, 1)$ và in ra giá trị của $\lambda_{100}$, $x_{100}$, $\lambda_{200}$, $x_{200}$.

```python
import numpy as np

A = np.array([
    [-4,  7,  8,  1],
    [-2, -1, -4,  9],
    [-4, -2,  6,  6],
    [ 9,  6,  9,  3]
], dtype=float)

x = np.array([1, 1, 1, 1], dtype=float)

for k in range(200):
    y = A @ x
    lam = np.linalg.norm(y, 2)
    x = y / lam
    
    if k == 99 or k == 199:
        print(f'Bước {k+1}: lam = {lam}, x = {x}')
```
