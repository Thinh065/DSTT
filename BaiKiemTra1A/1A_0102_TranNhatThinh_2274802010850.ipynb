## **Bài 1** *(7 điểm)*

Cho hệ phương trình tuyến tính như sau
$$\left\{\begin{array}{rcl}
x - 2y & = & 1 \\
3x + 2y & = & 11
\end{array}\right.$$

1.   Giải hệ phương trình trên sử dụng hàm `solve` của thư viện SymPy. *(2 điểm)*
2.   Giải hệ phương trình trên sử dụng phương pháp ma trận nghịch đảo. *(1 điểm)*
3.   Giải hệ phương trình trên sử dụng hàm `solve` của thư viện Numpy. *(1 điểm)*
4.   Giải hệ phương trình trên sử dụng hàm `solve` của thư viện SciPy. *(1 điểm)*
5.   Giải hệ phương trình trên sử dụng phương pháp Cramer. *(2 điểm)*

```python
x_sympy = None
y_sympy = None
print(f'Nghiệm theo SymPy                : x = {x_sympy}, y = {y_sympy}')

x_inv = None
y_inv = None
print(f'Nghiệm theo PP ma trận nghịch đảo: x = {x_inv}, y = {y_inv}')

x_np = None
y_np = None
print(f'Nghiệm theo hàm solve của Numpy  : x = {x_np}, y = {y_np}')

x_scipy = None
y_scipy = None
print(f'Nghiệm theo hàm solve của SciPy  : x = {x_scipy}, y = {y_scipy}')

x_cramer = None
y_cramer = None
print(f'Nghiệm theo phương pháp Cramer   : x = {x_cramer}, y = {y_cramer}')
```

```python
# 1. Giải hệ phương trình trên sử dụng hàm `solve` của thư viện SymPy. *(2 điểm)*
import sympy as sp

x, y = sp.symbols('x y')

sol = sp.solve([x - 2*y - 1, 3*x + 2*y - 11], (x, y))
x_sympy = sol[x]
y_sympy = sol[y]

print(f'Nghiệm theo SymPy                : x = {x_sympy}, y = {y_sympy}')
```

```python
# 2. Giải hệ phương trình trên sử dụng phương pháp ma trận nghịch đảo. *(1 điểm)*
import numpy as np

A = np.array([[1, -2],
              [3,  2]])
b = np.array([1, 11])

A_inv = np.linalg.inv(A)
result = A_inv.dot(b)

x_inv = result[0] # phần tử thứ 1 của vector nghiệm x
y_inv = result[1] # phần tử thứ 2 của vector nghiệm y

print(f'Nghiệm theo PP ma trận nghịch đảo: x = {x_inv}, y = {y_inv}')
```

```python
# 3. Giải hệ phương trình trên sử dụng hàm `solve` của thư viện Numpy. *(1 điểm)*
A = np.array([[1, -2],
              [3,  2]])
b = np.array([1, 11])

x_np, y_np = np.linalg.solve(A, b)

print(f'Nghiệm theo hàm solve của Numpy  : x = {x_np}, y = {y_np}')
```

```python
# 4. Giải hệ phương trình trên sử dụng hàm `solve` của thư viện SciPy. *(1 điểm)*
import numpy as np
from scipy import linalg

A = np.array([[1, -2],
              [3,  2]])
b = np.array([1, 11])

x_scipy, y_scipy = linalg.solve(A, b)

print(f'Nghiệm theo hàm solve của SciPy  : x = {x_scipy}, y = {y_scipy}')
```

```python
# 5. Giải hệ phương trình trên sử dụng phương pháp Cramer. *(2 điểm)*
A = np.array([[1, -2],
              [3,  2]])
b = np.array([1, 11]) 

detA = np.linalg.det(A)

Ax = np.array([[1, -2],
               [11, 2]])

Ay = np.array([[1, 1],
               [3, 11]])

x_cramer = np.linalg.det(Ax) / detA
y_cramer = np.linalg.det(Ay) / detA

print(f'Nghiệm theo phương pháp Cramer   : x = {x_cramer}, y = {y_cramer}')
```

## **Bài 2** *(3 điểm)*

Tìm giao điểm của ba đường thẳng $x - y = 2$, $2 x - y - z = 3$, và $x + y + z = 6$ trong không gian $\mathbb{R}^3$.

```python
x = None
y = None
z = None
print(f'Tọa độ giao điểm là: (x, y, z) = ({x}, {y}, {z})')
```

```python
import sympy as sp

x, y, z = sp.symbols('x y z')

# Hệ phương trình:
# x - y = 2
# 2x - y - z = 3
# x + y + z = 6

sol = sp.solve([x - y - 2, 2*x - y - z - 3, x + y + z - 6], (x, y, z))

x = sol[x]
y = sol[y]
z = sol[z]

print(f'Tọa độ giao điểm là: (x, y, z) = ({x}, {y}, {z})')
```
